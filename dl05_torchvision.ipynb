```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns

# ==========================================
# 1. 학습 장치(Device) 설정
# ==========================================
# 딥러닝 연산 속도를 극대화하기 위해 CUDA(GPU) 사용 가능 여부를 확인하고, 
# 사용 가능하면 GPU를, 그렇지 않으면 CPU를 기본 연산 장치로 지정합니다.
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")


# ==========================================
# 2. 전처리 정의 및 MNIST Dataset 로드
# ==========================================
# 이미지 데이터에 적용할 변환(Transform) 파이프라인을 정의합니다.
transform = transforms.Compose([
    transforms.ToTensor(),                          # ③ Tensor 변환: PIL Image 또는 NumPy 배열을 PyTorch Tensor로 변환하고 픽셀 값을 [0.0, 1.0] 범위로 스케일링합니다.
    transforms.Normalize((0.1307,), (0.3081,))      # 정규화: MNIST 데이터셋의 전체 평균(0.1307)과 표준편차(0.3081)를 이용해 데이터 분포를 조정하여 학습을 안정화합니다.
])

# ① MNIST Dataset 로드: torchvision.datasets을 이용해 MNIST 손글씨 데이터셋을 지정된 경로('./data')에 다운로드하고 로드합니다.
full_train_dataset = torchvision.datasets.MNIST(
    root='./data', 
    train=True, 
    download=True, 
    transform=transform
)

test_dataset = torchvision.datasets.MNIST(
    root='./data', 
    train=False, 
    download=True, 
    transform=transform
)


# ==========================================
# 3. 데이터셋 분할 및 DataLoader 구성
# ==========================================
# 모델의 과적합(Overfitting)을 방지하고 일반화 성능을 검증하기 위해 
# 전체 학습 데이터를 50,000개의 학습용과 10,000개의 검증(Validation)용 데이터셋으로 분할합니다.
train_size = 50000
val_size = 10000
train_dataset, val_dataset = torch.utils.data.random_split(
    full_train_dataset, [train_size, val_size]
)

# ④ DataLoader 구성: 대규모 데이터를 한 번에 처리하기 어려우므로, 
# 미니배치(Batch Size=64) 단위로 데이터를 묶어주고 셔플(Shuffle=True)을 통해 학습 순서를 무작위로 섞어 로드합니다.
train_loader = torch.utils.data.DataLoader(dataset=train_dataset, batch_size=64, shuffle=True)
val_loader = torch.utils.data.DataLoader(dataset=val_dataset, batch_size=1000, shuffle=False)
test_loader = torch.utils.data.DataLoader(dataset=test_dataset, batch_size=1000, shuffle=False)

# ② 이미지 데이터 확인: 로드된 데이터로더의 첫 번째 배치 샘플을 꺼내어 
# 입력 이미지 텐서와 정답 레이블 텐서의 형태(Shape)가 올바른지 직접 확인합니다.
examples = enumerate(train_loader)
batch_idx, (sample_data, sample_targets) = next(examples)
print(f"-> 입력 이미지 텐서 Shape: {sample_data.shape}")  # 예상 출력: torch.Size([64, 1, 28, 28])
print(f"-> 정답 레이블 텐서 Shape: {sample_targets.shape}") # 예상 출력: torch.Size([64])


# ==========================================
# 4. DNN 모델 구현 (Fully Connected Layers 구성)
# ==========================================
# ⑤ DNN 모델 구현: PyTorch의 nn.Module을 상속받아 다층 퍼셉트론(MLP/DNN) 구조를 정의합니다.
class DNN(nn.Module):
    def __init__(self):
        super(DNN, self).__init__()
        
        # ⑥ 첫 번째 Hidden Layer (Linear Layer) 구성: 입력 28x28(=784) 차원을 받아 256차원으로 변환하는 완전 연결 층입니다.
        self.fc1 = nn.Linear(28 * 28, 256)
        
        # ⑦ 두 번째 Hidden Layer (Linear Layer) 구성: 256차원의 특성을 받아 더 압축된 128차원으로 변환합니다.
        self.fc2 = nn.Linear(256, 128)
        
        # ⑧ Output Fully Connected Layer 구성: 최종 128차원을 받아 0부터 9까지의 클래스별 점수(Logits)를 출력하는 10차원으로 매핑합니다.
        self.fc3 = nn.Linear(128, 10)

    def forward(self, x):
        # DNN 입력을 위해 2D 이미지 텐서(Batch, 1, 28, 28)를 1차원 벡터 형태(Batch, 784)로 평탄화(Flatten)합니다.
        x = x.view(-1, 28 * 28)
        
        # 각 은닉층마다 선형 연산 후 비선형성(Non-linearity)을 추가하기 위해 ReLU 활성화 함수를 통과시킵니다.
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        
        # 최종 출력층은 활성화 함수를 거치지 않은 원시 점수(Logits) 형태로 반환합니다.
        x = self.fc3(x)
        return x

# 정의한 DNN 모델의 인스턴스를 생성하고 연산 장치(GPU 또는 CPU)로 이동시킵니다.
model = DNN().to(device)


# ==========================================
# 5. Loss Function 및 Optimizer 구성
# ==========================================
# ⑨ Loss Function 구성: 다중 클래스 분류(Multi-class Classification) 문제에서 
# 모델의 예측값과 실제 정답 간의 오차를 계산하기 위해 CrossEntropyLoss를 사용합니다.
criterion = nn.CrossEntropyLoss()                  

# ⑩ Optimizer 구성: 모델의 가중치를 업데이트할 최적화 알고리즘으로 Adam 옵티마이저를 선택하고, 학습률(Learning Rate)을 0.001로 설정합니다.
optimizer = optim.Adam(model.parameters(), lr=0.001)  


# ==========================================
# 6. Training Loop 및 Validation 구현
# ==========================================
num_epochs = 5
train_losses, val_losses = [], []
train_accuracies, val_accuracies = [], []

print("\n--- 모델 학습 및 검증 시작 ---")
for epoch in range(num_epochs):
    # ⑪ Training Loop 구현: 전체 데이터셋을 반복(Epoch)하며 미니배치 단위로 가중치를 학습합니다.
    model.train()  # 모델을 학습 모드로 설정 (Dropout, BatchNorm 등에 영향)
    running_loss = 0.0
    correct_train = 0
    total_train = 0
    
    for images, labels in train_loader:
        # 미니배치 데이터를 지정된 장치(GPU/CPU)로 전송합니다.
        images, labels = images.to(device), labels.to(device)
        
        # 이전 스텝에서 누적된 옵티마이저의 기울기(Gradient) 버퍼를 초기화합니다.
        optimizer.zero_grad()
        
        # 순전파(Forward Pass): 모델에 이미지를 입력하여 예측 결과(Logits)를 도출합니다.
        outputs = model(images)
        
        # 손실(Loss) 계산: 예측값과 실제 정답 레이블 간의 오차를 산출합니다.
        loss = criterion(outputs, labels)
        
        # 역전파(Backward Pass): 오차에 대한 각 가중치의 기울기를 미분을 통해 계산합니다.
        loss.backward()
        
        # 가중치 갱신: 계산된 기울기를 바탕으로 옵티마이저가 모델 파라미터를 업데이트합니다.
        optimizer.step()
        
        # 학습 손실 및 정확도 계산을 위한 누적 통계 수집
        running_loss += loss.item() * images.size(0)
        _, predicted = torch.max(outputs.data, 1)
        total_train += labels.size(0)
        correct_train += (predicted == labels).sum().item()
        
    epoch_train_loss = running_loss / len(train_loader.dataset)
    epoch_train_acc = 100 * correct_train / total_train  # ⑭ Accuracy 계산 (Train 정확도 산출)
    train_losses.append(epoch_train_loss)
    train_accuracies.append(epoch_train_acc)
    
    # ⑫ Validation 구현: 학습 도중 과적합 여부를 모니터링하기 위해 검증 데이터셋을 통한 평가를 수행합니다.
    model.eval()  # 모델을 평가 모드로 전환
    val_loss = 0.0
    correct_val = 0
    total_val = 0
    
    with torch.no_grad():  # 검증 과정에서는 기울기 계산이 불필요하므로 메모리 절약을 위해 비활성화합니다.
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            
            val_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs.data, 1)
            total_val += labels.size(0)
            correct_val += (predicted == labels).sum().item()
            
    epoch_val_loss = val_loss / len(val_loader.dataset)
    epoch_val_acc = 100 * correct_val / total_val  # ⑭ Accuracy 계산 (Validation 정확도 산출)
    val_losses.append(epoch_val_loss)
    val_accuracies.append(epoch_val_acc)
    
    print(f"Epoch [{epoch+1}/{num_epochs}] | "
          f"Train Loss: {epoch_train_loss:.4f}, Train Acc: {epoch_train_acc:.2f}% | "
          f"Val Loss: {epoch_val_loss:.4f}, Val Acc: {epoch_val_acc:.2f}%")


# ==========================================
# 7. Loss 시각화
# ==========================================
# ⑮ Loss 시각화: 학습 및 검증 과정에서 기록된 에폭별 손실(Loss) 변화 추이를 
# Matplotlib 라이브러리를 사용하여 선 그래프로 시각화하고 이미지 파일로 저장합니다.
plt.figure(figsize=(10, 5))
plt.plot(train_losses, label='Train Loss', marker='o')
plt.plot(val_losses, label='Validation Loss', marker='s')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss Curve (DNN)')
plt.legend()
plt.grid(True)
plt.savefig('dnn_loss_curve.png')
plt.show()


# ==========================================
# 8. Test 구현 및 Accuracy 계산
# ==========================================
# ⑬ Test 구현: 학습에 전혀 사용되지 않은 완전히 새로운 테스트 데이터셋을 통해 최종 성능을 평가합니다.
model.eval()
correct_test = 0
total_test = 0
all_preds = []
all_targets = []

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        
        # 가장 높은 확률(점수)을 가진 클래스 인덱스를 예측값으로 선택합니다.
        _, predicted = torch.max(outputs.data, 1)
        
        total_test += labels.size(0)
        correct_test += (predicted == labels).sum().item()  # ⑭ Accuracy 계산 (Test 최종 정확도 산출)
        
        # 이후 Confusion Matrix 및 오분류 분석을 위해 예측값과 정답 배열을 수집합니다.
        all_preds.extend(predicted.cpu().numpy())
        all_targets.extend(labels.cpu().numpy())

test_accuracy = 100 * correct_test / total_test
print(f"\n[Test 결과] 최종 테스트 정확도(Accuracy): {test_accuracy:.2f}%")


# ==========================================
# 9. Confusion Matrix 작성 및 시각화
# ==========================================
# ⑯ Confusion Matrix 작성: 실제 정답 클래스와 모델이 예측한 클래스 간의 혼동 행렬을 
# Scikit-Learn의 confusion_matrix로 계산한 후, Seaborn 히트맵으로 시각화합니다.
conf_matrix = confusion_matrix(all_targets, all_preds)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix (DNN)')
plt.savefig('dnn_confusion_matrix.png')
plt.show()


# ==========================================
# 10. 오분류 이미지 분석 (Misclassified Image Analysis)
# ==========================================
# ⑰ 오분류 이미지 분석: 테스트 데이터 중 모델이 틀리게 예측한 샘플들의 인덱스를 찾아내고, 
# 어떤 이미지를 어떤 숫자로 오인했는지 시각적으로 분석합니다.
all_preds_np = np.array(all_preds)
all_targets_np = np.array(all_targets)
misclassified_idx = np.where(all_preds_np != all_targets_np)[0]

print(f"\n[오분류 분석] 테스트 데이터 중 틀린 이미지 개수: {len(misclassified_idx)}개")

plt.figure(figsize=(12, 5))
for i, idx in enumerate(misclassified_idx[:5]):  # 오분류된 이미지 중 상위 5개를 시각화
    image, true_label = test_dataset[idx]
    pred_label = all_preds_np[idx]
    
    plt.subplot(1, 5, i + 1)
    # 정규화 과정에서 변형된 픽셀 값을 원래의 밝기 형태로 역변환하여 복원합니다.
    img_np = image.squeeze().numpy() * 0.3081 + 0.1307
    plt.imshow(img_np, cmap='gray')
    plt.title(f"True: {true_label}\nPred: {pred_label}")
    plt.axis('off')
plt.tight_layout()
plt.savefig('dnn_misclassified_images.png')
plt.show()


# ==========================================
# 11. 모델 저장 (Model Saving)
# ==========================================
# ⑱ 모델 저장: 학습이 완료된 DNN 모델의 가중치 상태(state_dict)를 
# 추후 재사용 및 배포를 위해 파일('mnist_dnn_model.pth')로 안전하게 저장합니다.
torch.save(model.state_dict(), 'mnist_dnn_model.pth')
print("\n[모델 저장 완료] 학습된 DNN 모델 가중치가 'mnist_dnn_model.pth'로 저장되었습니다.")
```
